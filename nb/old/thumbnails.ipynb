```python
%load_ext autoreload
%autoreload 2
import numpy as np
import matplotlib.pyplot as plt

import comborun

from config import generate_full_paths, ps_data_ranges
```

```python

group = 'ps20nm'

cbr1 = comborun.ComboRun(generate_full_paths(ps_data_ranges[group], 25))
thumbnails = cbr1.get_thumbnails(32)
fig, axes = plt.subplots(4,4)

np.random.seed(15)

for i in range(16):
    thu_i = np.random.randint(thumbnails.shape[0])
    axes.flatten()[i].imshow(thumbnails[thu_i,...], clim=(0,100))
    axes.flatten()[i].set_title(thu_i)


np.sum(thumbnails)
print(f'{group} unfocused+focused')
```

```python
cbr1 = comborun.ComboRun(generate_full_paths(ps_data_ranges[group], 25))
cbr2 = comborun.ComboRun(generate_full_paths(ps_data_ranges[group], 7))

cbr1.filter_focused(cbr2)
cbr1.scatter_xy()
```

```python
cbr1 = comborun.ComboRun(generate_full_paths(ps_data_ranges[group], 25))
cbr2 = comborun.ComboRun(generate_full_paths(ps_data_ranges[group], 7))

cbr1.filter_focused(cbr2)
fig, axes = plt.subplots(4,4)

np.random.seed(20)

for i in range(16):
    thu_i = np.random.choice(np.where(cbr1.filter)[0])
    axes.flatten()[i].imshow(thumbnails[thu_i,...], clim=(0,100))
    axes.flatten()[i].set_title(thu_i)


np.sum(thumbnails)
print(f'{group} focused')
```

```python
all_thbn = []
for group in ps_data_ranges.keys():
    
    cbr = comborun.ComboRun(generate_full_paths(ps_data_ranges[group], 25))
    print(f'{group}: with {cbr.peak_xs.size} peaks.')
    cbr_thbn = cbr.get_thumbnails()
    all_thbn.append(cbr_thbn)

result = np.concatenate(all_thbn, axis=0)
np.save(f'{config.DATA_DIR}/thumbnails.npy',result)
```
